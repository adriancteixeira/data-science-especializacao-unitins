## 1.1 Download e carregamento dos arquivos

```python
from google.colab import drive
import os
import requests
import pandas as pd

# 1.1.1 Montar o Drive
drive.mount('/content/drive')

# 1.1.2 Arquivos a baixar: (url, caminho_destino)
pasta = '/content/drive/MyDrive/Pós Graduação - Aulas/Aspectos Avançados de Python/Arquivos/Atividades'

arquivos = [
    ('http://www.unitins.br/arquivos/atividade_alunos.csv',      f'{pasta}/atividade_alunos.csv'),
    ('http://www.unitins.br/arquivos/atividade_desempenho.csv',  f'{pasta}/atividade_desempenho.csv'),
]

# 1.1.3 Criar pasta e baixar apenas os que ainda não existem
os.makedirs(pasta, exist_ok=True)

for url, copia in arquivos:
    if not os.path.exists(copia):
        response = requests.get(url)
        response.raise_for_status()
        with open(copia, 'wb') as f:
            f.write(response.content)
        print(f'{os.path.basename(copia)} baixado.')
    else:
        print(f'{os.path.basename(copia)} já existe.')

# 1.1.4 Carregar os arquivos
df_alunos = pd.read_csv(f'{pasta}/atividade_alunos.csv')
df_desempenho = pd.read_csv(f'{pasta}/atividade_desempenho.csv')

print('\nAlunos:')
display(df_alunos.head(5))

print('\nDesempenho:')
display(df_desempenho.head(5))
```

## 1.2 Verificação dos dados e documentação das inconsistências

```python
# 1.2.1 Verificação do shape
print("--- Shape do dataframe 'Alunos' ---")
display(df_alunos.shape)

print("\n--- Shape do dataframe 'Desempenho' ---")
display(df_desempenho.shape)



# 1.2.2 Identificar valores faltantes
print("\n--- Contagem de valores nulos por coluna e percentual do dataframe 'Alunos' ---")
display(df_alunos.isnull().sum())
display(df_alunos.isnull().sum() / len(df_alunos) * 100)

print("\n--- Contagem de valores nulos por coluna e percentual do dataframe 'Desempenho' ---")
display(df_desempenho.isnull().sum())
display(df_desempenho.isnull().sum() / len(df_desempenho) * 100)



# 1.2.3 Identificar tipos de dados
print("\n--- Tipos de dados do dataframe 'Alunos' ---")
display(df_alunos.dtypes)

print("\n--- Tipos de dados do dataframe 'Desempenho' ---")
display(df_desempenho.dtypes)



# 1.2.4 Identificar inconsistências
print("\n--- Valores únicos da coluna 'sexo' ---")
display(df_alunos['sexo'].unique())

print("\n--- Valores únicos da coluna 'cidade_polo' ---")
display(df_alunos['cidade_polo'].unique())

print("\n--- Valores únicos da coluna 'curso' ---")
display(df_alunos['curso'].unique())

print("\n--- Valores únicos da coluna 'disciplina' ---")
display(df_desempenho['disciplina'].unique())



# 1.2.5 Visualizar linhas com pelo menos um NaN
df_alunos_com_nan = df_alunos[df_alunos.isnull().any(axis=1)]
print("\n--- Todas as linhas contendo NaN do dataframe 'Alunos' ---")
display(df_alunos_com_nan)

df_desempenho_com_nan = df_desempenho[df_desempenho.isnull().any(axis=1)]
print("\n--- Todas as linhas contendo NaN do dataframe 'Desempenho' ---")
display(df_desempenho_com_nan)
```

Na etapa de análise, foi possível verificar que a base de dados `alunos.csv` possui 11 linhas e 8 colunas. Já a base de dados `desempenho.csv` possui 283 linhas e 6 colunas. Como 283 não é divisível por 11, é possível rapidamente concluir que não há o mesmo número de informações de desempenho para todos os alunos.

Não há valores nulos na base de dados para os alunos, mas há 26 valores nulos na média na base de dados de desempenho, representando aproximadamente 9,2% do total de dados.

A coluna `id_aluno` em ambos os dataframes está categorizada como `int64`, o que pode causar problemas pois será tratada como um número inteiro (e não como um dado fixo, como deve ser) dependendo da manipulação feita nos dados.

## 2. Limpeza dos dados

Padronize a coluna sexo para os valores 'Masculino' e 'Feminino'.
Corrija e padronize a coluna `cidade_polo`.
Substitua os valores vazios da coluna `media` por 0.
Remova registros duplicados caso existam.

```python
# 2.1 Padronização da coluna sexo
mapa_sexo = {
    'M': 'Masculino',
    'F': 'Feminino',
    'Feminino': 'Feminino',
    'Masculino': 'Masculino',
}

df_alunos['sexo'] = df_alunos['sexo'].map(mapa_sexo)

# 2.2 Correção e padronização da coluna cidade_polo
df_alunos['cidade_polo'] = df_alunos['cidade_polo'].str.strip()

mapa_cidade_polo = {
    'COLMÉIA': 'COLMÉIA',
    'PALMAS-TOCANTINS': 'PALMAS',
    'MIRANORTE-TO': 'MIRANORTE',
    'PALMAS': 'PALMAS',
    'Palmas-TO': 'PALMAS',
    'COLINAS': 'COLINAS'
}

df_alunos['cidade_polo'] = df_alunos['cidade_polo'].map(mapa_cidade_polo)

# 2.3 Substituindo os valores vazios (NaN) da coluna media por 0
df_desempenho['media'] = df_desempenho['media'].fillna(0)

# 2.4 Removendo registros duplicados
df_alunos = df_alunos.drop_duplicates()
df_desempenho = df_desempenho.drop_duplicates()

# 2.5 Convertendo coluna id_aluno de int64 para string
df_alunos['id_aluno'] = df_alunos['id_aluno'].astype('string')
df_desempenho['id_aluno'] = df_desempenho['id_aluno'].astype('string')

# 2.6 Conferência dos resultados
print("\n--- Quantidade de valores da coluna 'sexo' ---")
display(df_alunos['sexo'].value_counts(dropna=False))

print("\n--- Quantidade de valores da coluna 'cidade_polo' ---")
display(df_alunos['cidade_polo'].value_counts(dropna=False))

print("\n--- Valores nulos do dataframe 'alunos' ---")
display(df_alunos.isnull().sum())

print("\n--- Valores nulos do dataframe 'desempenho' ---")
display(df_desempenho.isnull().sum())
```

Conforme as orientações do projeto e outros problemas percebidos, as seguintes alterações foram realizadas:

*   A coluna `sexo` foi padronizada para 'Masculino' e Feminino';
*   A coluna `cidade_polo` foi padronizada para manter apenas o nome da cidade, sem o estado, em letras maiúsculas como está a maioria do restante dos dados da tabela. Como as cidades são todas do estado do Tocantins, considerou-se irrevelante mencionar o estado;
*   Os valores NaN da coluna `media` foram substituídos por zero;
*   A coluna `id_aluno` foi convertida de int64 para string, pois é um dado que não deve ser tratado como número inteiro.


## 3. Integração

Realize o merge entre as duas tabelas utilizando inner join pelo campo id_aluno.

```python
# 3.1 Inner join
df_total = pd.merge(df_alunos, df_desempenho, on='id_aluno', how='inner')

# 3.2 Visualização do dataframe após a junção
print("\n--- Shape do dataframe 'Total' ---")
display(df_total.shape)
display(df_total.head())
```

Após a junção dos dois dataframes, verifica-se que a quantidade de linhas permanece a mesma da base de dados `desempenho.csv`, enquanto a quantidade de colunas se expandiu e agora inclui os dados de ambas as bases de dados, como podemos verificar na tabela.

Isso indica um `merge()` feito com sucesso, e agora os dados podem seguir para a etapa de análise.
